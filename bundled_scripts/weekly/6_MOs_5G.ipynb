```python
# FunÃƒÆ’Ã‚Â§ÃƒÆ’Ã‚Â£o para carregar os arquivos
base_path = r"C:\Dumps\todos_enms"

def load_parquet(pattern, columns=None):
    import glob, pandas as pd
    files = glob.glob(pattern)
    if not files:
        return pd.DataFrame()
    df = pd.concat((pd.read_parquet(f, columns=columns) for f in files),
                   ignore_index=True)
    try:
        df = df.convert_dtypes(dtype_backend="pyarrow")
    except Exception:
        pass
    return df
```

```python
import os
import glob
import pandas as pd
import numpy as np

# Base onde o parser gerou os .parquet

pattern = fr"{base_path}\ExternalNRCellCU_*_part*.parquet"

# Colunas realmente usadas (mais rÃƒÆ’Ã‚Â¡pido ler sÃƒÆ’Ã‚Â³ elas)
cols_needed = ["eNB", "ExternalGNBCUCPFunction", "cellLocalId", "nRFrequencyRef", "externalNRCellCUId", "nRPCI"]

# Loader parquet
files = glob.glob(pattern)
if not files:
    print(f"Nenhum parquet encontrado para ExternalNRCellCU: {pattern}")
    df1 = pd.DataFrame()
else:
    df1 = pd.concat((pd.read_parquet(f, columns=cols_needed) for f in files),
                    ignore_index=True)
    # opcional: economizar RAM (strings Arrow). Se nÃƒÆ’Ã‚Â£o quiser Arrow, remova o try/except.
    try:
        df1 = df1.convert_dtypes(dtype_backend="pyarrow")
    except Exception:
        pass

if not df1.empty:
    try:
        # eNB -> gNB e UF
        df1.rename(columns={'eNB': 'gNB'}, inplace=True)
        df1['UF'] = df1['gNB'].astype("string").str[-2:]

        # Normaliza ExternalGNBCUCPFunction ÃƒÂ¢Ã¢â‚¬Â Ã¢â‚¬â„¢ CGI base
        def alterar_valor(valor):
            try:
                if isinstance(valor, str):
                    if valor.startswith('auto'):
                        valor = valor[4:]                      # remove 'auto'
                        partes = valor.split('_')
                        # junta tudo menos os 2 ÃƒÆ’Ã‚Âºltimos + '-' + ÃƒÆ’Ã‚Âºltimo
                        valor = ''.join(partes[:-2]) + '-' + partes[-1]
                        valor = valor.split(',')[0]
                    elif not valor.startswith('724'):
                        valor = valor.split(',')[0]
                return valor
            except Exception:
                return valor

        df1['CGI'] = df1['ExternalGNBCUCPFunction'].apply(alterar_valor)

        # Se houver cellLocalId, concatena ao CGI (com '-')
        if 'cellLocalId' in df1.columns:
            df1['CGI'] = np.where(
                df1['cellLocalId'].notna(),
                df1['CGI'].astype("string") + '-' + pd.to_numeric(df1['cellLocalId'], errors='coerce').astype('Int64').astype('string'),
                df1['CGI'].astype("string")
            )
        else:
            print("Coluna 'cellLocalId' nÃƒÆ’Ã‚Â£o encontrada. Pulando concatenaÃƒÆ’Ã‚Â§ÃƒÆ’Ã‚Â£o com CGI.")

        # Extrai nRFrequency de nRFrequencyRef (evita erro no Arrow: grupo nomeado)
        df1['nRFrequency'] = df1['nRFrequencyRef'].astype("string").str.extract(
            r"NRFrequency=(?P<val>[^,]+)", expand=False
        )

        # Faxina geral
        df1 = df1.dropna(thresh=1, axis='columns')
        df1['ExternalGNBCUCPFunction'] = df1['ExternalGNBCUCPFunction'].astype("string").str.split(',').str[0]
        # Salva TXT final
        out_dir = r"C:\Dumps\LTE_Total"
        os.makedirs(out_dir, exist_ok=True)
        out_path = os.path.join(out_dir, "5G_ExternalNRCellCU.txt")

        df1.to_csv(out_path, sep='\t', index=False, encoding='utf-8', lineterminator='\n', na_rep='')
        print(f"5G_ExternalNRCellCU OK ÃƒÂ¢Ã¢â‚¬Â Ã¢â‚¬â„¢ {out_path} ({len(df1):,} linhas, {df1.shape[1]} colunas)")
    except Exception as e:
        print(f"Erro durante o processamento do DataFrame: {e}")
else:
    print("DataFrame df1 estÃƒÆ’Ã‚Â¡ vazio. Pulando etapa de processamento...")
```
